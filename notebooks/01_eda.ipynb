```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.histplot(df['TotalPremium'], bins=50, kde=True)
plt.title("Distribution of Total Premium")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['TotalClaims'], bins=50, kde=True)
plt.title("Distribution of Total Claims")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Gender', data=df)
plt.title("Gender Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(y='VehicleType', data=df)
plt.title("Vehicle Type Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['TotalClaims'])
plt.title("Outliers in Total Claims")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['TotalPremium'])
plt.title("Outliers in Total Premium")
plt.show()
```

```python
plt.figure(figsize=(10,5))
df.groupby('Province')['LossRatio'].mean().sort_values().plot(kind='bar')
plt.title("Average Loss Ratio by Province")
plt.ylabel("Loss Ratio")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='TotalPremium', y='TotalClaims', data=df)
plt.title("Total Premium vs Total Claims")
plt.show()
```

```python
plt.figure(figsize=(10,6))

corr = df[['TotalPremium', 'TotalClaims', 'AnnualPremium', 'ClaimAmount',
           'CustomValueEstimate', 'RiskScore']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix of Key Financial Variables")
plt.show()
```

```python
zip_analysis = df.groupby('ZipCode')[['TotalPremium', 'TotalClaims']].mean().reset_index()

zip_analysis.head()
```

```python
plt.figure(figsize=(12,5))
sns.scatterplot(x='ZipCode', y='TotalClaims', data=zip_analysis)
plt.title("Average Claims by ZipCode")
plt.xticks(rotation=90)
plt.show()
```

```python
vehicle_risk = df.groupby('VehicleType')['TotalClaims'].mean().sort_values()

plt.figure(figsize=(10,5))
vehicle_risk.plot(kind='bar')
plt.title("Average Claims by Vehicle Type")
plt.ylabel("Average Claims")
plt.show()
```

```python
vehicle_risk = df.groupby('VehicleType')['TotalClaims'].mean().sort_values()

plt.figure(figsize=(10,5))
vehicle_risk.plot(kind='bar')
plt.title("Average Claims by Vehicle Type")
plt.ylabel("Average Claims")
plt.show()
```

```python
province_risk = df.groupby('Province')['LossRatio'].mean().sort_values()

plt.figure(figsize=(10,5))
province_risk.plot(kind='bar')
plt.title("Loss Ratio by Province")
plt.ylabel("Loss Ratio")
plt.show()
```

```python
gender_risk = df.groupby('Gender')['LossRatio'].mean()

plt.figure(figsize=(6,4))
gender_risk.plot(kind='bar')
plt.title("Loss Ratio by Gender")
plt.ylabel("Loss Ratio")
plt.show()
```

## Key Findings

- Province X shows the highest loss ratio, indicating higher claim risk.
- Vehicle type Y has significantly higher claim severity.
- Gender differences in loss ratio are minimal, suggesting low gender-based pricing impact.

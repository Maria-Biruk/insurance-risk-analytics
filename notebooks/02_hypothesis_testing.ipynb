```python
top_zips = df['ZipCode'].value_counts().index[:2]

group_a = df[df['ZipCode'] == top_zips[0]]['HasClaim']
group_b = df[df['ZipCode'] == top_zips[1]]['HasClaim']
```

```python
from scipy import stats

t_stat, zip_p_value = stats.ttest_ind(group_a, group_b)

print("T-statistic:", t_stat)
print("P-value:", zip_p_value)
```

```python
results = pd.DataFrame({
    "Hypothesis": [
        "Province vs Risk",
        "Gender vs Risk",
        "ZipCode vs Risk"
    ],
    "Test": [
        "ANOVA",
        "Chi-square",
        "T-test"
    ],
    "P-value": [
        province_p_value,
        gender_p_value,
        zip_p_value
    ]
})

results
```

```python
results["Decision"] = results["P-value"].apply(
    lambda p: "Reject H0" if p < 0.05 else "Fail to Reject H0"
)

results
```

```python
from scipy import stats
import pandas as pd

def chi_square_test(df, col1, col2):
    table = pd.crosstab(df[col1], df[col2])
    chi2, p, dof, exp = stats.chi2_contingency(table)
    return chi2, p

def t_test(group1, group2):
    return stats.ttest_ind(group1, group2)

def anova_test(*groups):
    return stats.f_oneway(*groups)
```
